```python
def gc_content(seq):
    seq = seq.upper()
    gc = seq.count("G") + seq.count("C")
    return gc / len(seq) * 100

print(gc_content("ATGCGATGCGCT"))
print(f"{gc_content('atgc'):.2f}")
```

```python
def gc_content(seq, decimals=2):
    seq = seq.upper()
    gc = seq.count("G") + seq.count("C")
    return round(gc / len(seq) * 100, decimals)

print(gc_content("ATGCGCGATTATAT"))
print(gc_content("ATGCGCGATATAAT", 0))
assert gc_content("ATGC") == 50.0
assert gc_content("GGCC") == 100.0
assert gc_content("ATAT") == 0.0
print("all tests passed")
```

```python
def count_base(seq , base):
  return seq.upper().count(base.upper())
print(count_base("ATGAT", "A"))
print(count_base("atgat", "a"))
print(count_base("ATG", "C"))
assert count_base("ATGAT", "A") == 2
assert count_base("atgat", "a") == 2
assert count_base("ATG", "C") == 0
print("all tests passed")
```

```python
def is_valid(seq):
  seq = seq.upper()
  for base in seq:
      if base not in "ATGC":
        return False
  return True
print(is_valid("ATGCGCGATTATAT"))
print(is_valid("thbhgad"))
```

```python

def primer_tm(primer):
  at = count_base(primer,"A") + count_base(primer,"T")
  gc = count_base(primer,"C") + count_base(primer,"G")

  return 2*at + 4*gc
print(primer_tm("AATTGCGTA"))
```

```python
def split_codons(seq):
    seq = seq.upper()
    codons = []
    for i in range(0, len(seq), 3):
        codon = seq[i:i+3]
        if len(codon) == 3:
            codons.append(codon)
    return codons

print(split_codons("ATGGCCAAATTTGGGTAA"))
print(split_codons("ATGGCCAAATTTGGGTAAC"))
```
